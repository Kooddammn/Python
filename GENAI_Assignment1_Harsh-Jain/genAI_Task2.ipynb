```python
#created a set of categories
categories = {"KitchenWare","Electronics","Clothing","Fashion","Medicinal","Gaming","Household","Sports"}
```

```python
#Demonstrated adding new category to the set
categories.add("Educational")
categories.add("Automobile")
#Duplicate record are ignored it already present in the set, for ex kitchenware
categories.add("KitchenWare")
print(categories)
```

```python
#check whether category exists in set and print booleans result
if "Gaming" in categories:
    print(True)
else:
    print(False)
```

```python
#get total number of unique categories using set
print(len(categories))
```
