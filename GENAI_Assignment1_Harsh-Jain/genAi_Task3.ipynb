```python
#created dictionary named price_dict with product name and the price
price_dict ={
    "laptop":50000,
    "keyboard": 1000,
    "mouse": 500,
    "monitor": 10000,
    "headphone": 1200,
    "phone": 45000
}
```

```python
#Add new product with the price in price_dict
price_dict["tablet"] = 3000
#update the price of existing product
price_dict["headphone"] = 1500
#Remove the product by name
if "mouse" in price_dict:
 del price_dict["mouse"]
print(price_dict)
```

```python
#print average price of all product
sum = 0
for item in price_dict.values():
 sum += item
avg = sum/len(price_dict)
print(avg)
```

```python
#print the product with both maximum price and minimum price
product = max(price_dict.values()) * min(price_dict.values()) 
print(product)
```
