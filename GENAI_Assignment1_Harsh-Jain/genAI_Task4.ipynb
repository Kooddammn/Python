```python
#created list of tuple named catalog where each tuple is (product name, price, category)
catalog = [("cooking pan",3000,"kitchenware"),("grinder",4000,"kitchenware"),("laptop",50000,"Electronics"),("shirt",1200,"Clothing"),
           ("Perfume",400,"fashion"),("cricket bat",2000,"sports"),("football",1200,"sports"),("car",120000,"automobile"),("two wheeler",60000,"automobile")]
```

```python
#created dictionary which map each category to the list of product list
category_to_product = {}
for product_name, price, category in catalog:
    category_to_product.setdefault(category, []).append(product_name)
print(category_to_product)
```

```python
#print all product that belong to category which has maximum products
max_category = max(category_to_product, key=lambda k: len(category_to_product[k]))
for product in category_to_product[max_category]:
    print(product)
```
