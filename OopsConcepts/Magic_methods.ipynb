#### Magic method are dunder method because ther are declared using double underscore, that comes before and after the name

```python
class Avengers:
    def __init__(self,name,strenght):
        self.name = name
        self.strenght = strenght
    def __add__(self,other):
        return Avengers(self.name + " & "  + other.name,self.strenght + other.strenght)
    def __str__(self):
        return f"{self.name} with strenght {self.strenght}"
ironname = Avengers("Iron Man",60)
thor = Avengers("Thor",90)
team = ironname + thor
print(team)
```
