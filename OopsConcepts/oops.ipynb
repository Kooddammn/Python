```python
class Cartoon:
    def __init__(self,name,show,catchpharse):
        self.name = name
        self.show = show
        self.catchpharse = catchpharse
    
    def introduction(self):
        print(f"Hello my name is {self.name}")
    def speak(self):
        print(f"{self.name} says {self.catchpharse}")
    
    
```

```python
mickey_mouse = Cartoon("Mickey Mouse","Mickey Mouse club house","Oh! boy")
mickey_mouse.introduction()
mickey_mouse.speak()
print(mickey_mouse.name)
print(mickey_mouse.catchpharse)
print(mickey_mouse.show)
```

```python
spongebob = Cartoon("SpongeBob","Sponge Bob Square Pant","I 'm ready")
spongebob.introduction()
spongebob.speak()
```

```python
class Avengers:
    origin = "Avenger"
    def __init__(self,name,power):
        self.name = name
        self.power = power
    
    @classmethod
    def oirgin_info(cls):
        print(f"{cls.origin} assemble")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")
    
ironman = Avengers("Iron Man","Iron suite")
thor = Avengers("Thor","God of thunder")
ironman.self_info()
thor.self_info()
Avengers.oirgin_info()
```
