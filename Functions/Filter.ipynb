## Filter function is going to retrive only those  item which are meeting specific condition in an iterable


### The filter function constructs an Iterator from element of an iterable for which condition remains true or function returns true

#### Syntax for filter function

#### list(filter(func,iterable)))

```python
def even_function(num):
        return num % 2 ==0
iterable_ietm = [1,3,7,2,0,6,]
print(list(filter(even_function,iterable_ietm)))
```
