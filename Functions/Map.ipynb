## Map function applies a given function opreation to all the item given in the iterables or list and returns a map object

## Map functiona is particularly useful for transforming the data or applying an operation on all item of the iterables

## Syntax for map
## list(map(function, iterable))

```python
## square all item of the list
def square(list_item):
    empty_list = []
    for item in list_item:
        squared_item = item**2
        empty_list.append(squared_item)
    print(empty_list)

square([1,4,5,3,7,9])
```

```python
## function implemenation using map()
result_list = list(map(lambda x:x**2,[1,4,5,3,7,9]))
print(result_list)

lambda a,b,c,d,e : a**2 
```

```python
#map function with basic function in it
def square(num):
    return num**2
print(list(map(square,[1,3,5,7,4])))
```

```python
#Map value with multiple iterable
numbers_1 = [2,3,4,5]
numbers_2 = [1,4,6,2]

print(list(map(lambda x,y : x + y,numbers_1, numbers_2)))
```

```python
#Map() function to convert all string value in iterable with integer value
print(list(map(int,['1','4','7','5','9'])))
```

```python
employee_name = ['asd','hdg','uyr','tfd']
print(list(map(str.upper,employee_name)))
```

```python
#Retrives values inside dictionary using Map()

employees_details = [{'Name':'John','Age':26,'Department':'Business'},
                     {'Name':'Mike','Age':28, 'Department':'Sale'},
                     {'Name':'Tom','Age':29, 'Department':'Security'}]

def age_Retrival(list_item):
    return list_item['Age']
print(list(map(age_Retrival,employees_details)))
```
