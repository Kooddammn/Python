```python
#Task 1: Line Plot(Sales Trend)
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('sales_data_sample.csv')
plt.plot(df.MONTH_ID[1:100],df.SALES[1:100])
plt.title('Sales Trend')
plt.xlabel('month')
plt.ylabel('sales')
plt.show()
```

```python
#Task 2: Scatter PLot
plt.scatter(df.MONTH_ID[1:100],df.SALES[1:100])
plt.title("Scatter PLot")
plt.xlabel('Months')
plt.ylabel('Sales')
plt.show()
```

```python
#Task 3: Bar PLot
plt.bar(df.MONTH_ID[1:100], df.SALES[1:100])
plt.title("Vertical bar chart")
plt.ylabel('Sales')
plt.xlabel('Months')
plt.show()
plt.barh(df.MONTH_ID[1:100],df.SALES[1:100])
plt.title("Horizontal bar chart")
plt.ylabel('Months')
plt.xlabel('Sales')
plt.show()
```

```python
#Task 4: Multi Bar Plot
import matplotlib.pyplot as plt
import numpy as np

years = df.YEAR_ID[1:50]
sales = df.SALES[1:50]
salesB = df.SALES_B[1:50]

x = np.arange(len(years))

# Bar width
width = 0.5

# Multiple bars
plt.bar(x - width, sales, width, label='SALES A')
plt.bar(x , salesB, width, label='SALES B')

# Labels and title
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Sales Comparison Across Different Years')


# Legend
plt.legend()

# Display chart
plt.tight_layout()
plt.show()
```

```python
#Task 5: Stacked Bar chart
import matplotlib.pyplot as plt

# Data
years = df.YEAR_ID[1:20]
sales = df.SALES[1:20]
salesB = df.SALES_B[1:20]

# Stacked bar chart
plt.bar(years, sales, label='SALES A')
plt.bar(years, salesB, bottom=sales, label='SALES B')


# Labels and title
plt.xlabel('Year')
plt.ylabel('Sales')
plt.title('Stacked Sales by Year')

# Legend
plt.legend()

# Display chart
plt.tight_layout()
plt.show()
```

```python
#Task 6: Histogram
import matplotlib.pyplot as plt

countries = df.COUNTRY
plt.Figure(figsize=(10,10))

# Create histogram
plt.hist(countries, bins=12, edgecolor='black')

# Title and axis labels
plt.title('Distribution of Numeric Data')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Display chart
plt.tight_layout()
plt.show()
```

```python
#Task 7: Pie chart
import matplotlib.pyplot as plt

# Data
years = df.YEAR_ID[1:25]
sales = df.SALES[1:25]

# Create pie chart
plt.pie(
    sales,
    labels=years,
    autopct='%1.1f%%',
    startangle=90
)

# Title
plt.title('Market Share by Category')

# Display chart
plt.show()
```
