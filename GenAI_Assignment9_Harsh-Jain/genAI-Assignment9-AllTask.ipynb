```python
#Task1: Creating NumPy Arrays
#1.import numpy as ip
import numpy as np

#2. 1D of integer from 1 to 10
oneDim = np.arange(1,11)
print(oneDim)
print(f"Shape:{oneDim.shape}")
print(f"Data type: {oneDim.dtype}")
# 2D of shape (3,3) with value 1 to 9

twoDim = np.arange(1,10).reshape(3,3)
print(twoDim)
print(f"Shape:{twoDim.shape}")
print(f"Data type:{twoDim.dtype}")

# Numpy array from the list
array = np.array([10,20,30,40,50])
print(array)
print(f"Shape:{array.shape}")
print(f"Data type:{array.dtype}")
```

```python
#Task 2: Important Mathematical Operations
A = np.array([10,20,30,40])
B = np.array([1,2,3,4])
print(A+B)
print(A - B)
print(A*B)
print(A/B)
print(A**2)
```

```python
#Task 3: Importand NumPy Mathematical Formula
values = np.array([2,4,6,8,10])
print(np.sqrt(values))
print(np.exp(values))
print(np.log(values))
print(np.sum(values))
print(np.cumsum(values))
```

```python
#Task 4: Aggregation Operation
data = np.array([[10,20,30],
                 [40,50,60],
                 [70,80,90]])
print(np.sum(data, axis = 0))
print(np.sum(data, axis = 1))
print(np.min(data))
print(np.max(data))
print(np.mean(data))
```

```python
#Task 5: Statistical Operation (Core Focus)
marks = np.array([78, 85,90,66,72,88,95,60])
print(np.mean(marks))
print(np.median(marks))
print(np.var(marks))
print(np.std(marks))
print(np.min(marks))
print(np.max(marks))
print(np.ptp(marks))
```

```python
#Task 6: Percentiles and sorting
sort = np.sort(marks)
print(sort)
print(np.percentile(marks,25))
print(np.percentile(marks,50))
print(np.percentile(marks,75))
print(sort[ sort > np.average(marks)].size)
```

```python
#Task 7: Mini Use Case: Sales Analysis
sales = np.array([1200,1500,900,2000,1800,1700,1600])
print(f"Total weekly sales: {np.sum(sales)}")
print(f"Average Daily Sales: {np.average(sales)}")
print(f"Highest sale:{np.max(sales)} Lowest sales: {np.min(sales)}")
print(f"Standard deviation of sales: {np.std(sales)}")
print(np.argwhere(sales > np.average(sales)))
```
