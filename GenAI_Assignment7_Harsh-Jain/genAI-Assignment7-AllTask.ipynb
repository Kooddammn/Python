```python
#Task 1: Basic Class and object creation
#Task 2: Constructor and Encapsulation
#Task 6: Magic Method and operator overloading
class Product:
    def __init__(self, name, price, category):
        self.name = name
        self.__price = price
        self.category = category
    
    def get_info(self):
        print(f"Product Name: {self.name}, Product price: {self.__price}, Product category: {self.category}")
    
    def get_price(self):
        return self.__price

    def set_price(self,new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            raise ValueError("Price cannot be negative")
    
    def __str__(self):
        return f"Product name: {self.name}, Price: {self.__price}, Category: {self.category}"
    
    def __add__(self,other):
        return Product(self.name +" "+ other.name, int(self.__price) + int(other.__price),self.category +" "+ other.category)

product1 = Product("shoes",3000,"Fashion")
product1.get_price()
product1.set_price(5000)
product1.get_info()
product2 = Product("football",2000,"Sports")
product2.get_price()
product2.set_price(4000)
product2.get_info()
print(product1)
print(product2)
print(product1 + product2)

    
```

```python
#Task 3: Inheritance( Single level)
class ElectronicProduct(Product):
    def __init__(self,name,price,category,warranty_years):
        super().__init__(name,price,category)
        self.warranty_years = warranty_years
    def get_info(self):
        parent_info =  super().get_info()
        print(f"{parent_info} and warrantly year info : {self.warranty_years} ")

desk = Product("desk",15000,"product")
desk.get_info()
laptop = ElectronicProduct("laptop",60000,"electrinics",3)
laptop.get_info()

```

```python
#Task 4: Polymorphism
class Laptop(Product):
    def __init__(self,name,price,category):
        super().__init__(name,price,category)
    
    def get_info(self):
        return super().get_info()

class Mobile(Product):
    def __init__(self,name,price,category):
        super().__init__(name,price,category)
    
    def get_info(self):
        return super().get_info()
    
laptops = [
    Laptop("lenovo","30000","I3 procesor"),
    Laptop("dell","40000","I5 processor"),
    Laptop("acer","50000","I7 processor")
]

mobiles = [
    Mobile("samsung","70000","galaxy s25"),
    Mobile("apple","90000","iphone 17 pro max"),
    Mobile("google","80000","pixel 10")
]

for obj in laptops:
    obj.get_info()
for obj in mobiles:
    obj.get_info()
```

```python
#Task 5: Abstraction
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def process_payment(self,amount):
        pass
class CreditCardPayment(Payment):
    def __init__(self,amount):
        self.amount = amount
    def process_payment(self):
        print(f"Credit Card Payment {self.amount}")
class UPIPayment(Payment):
    def __init__(self,amount):
        self.amount = amount
    def process_payment(self):
        print(f"UPI Payment {self.amount}")

creditCardPayment = CreditCardPayment(15000)
upipayment = UPIPayment(20000)
creditCardPayment.process_payment()
upipayment.process_payment()
```

```python
#Task 7: Mini project simple inventory system
class Inventory:
    def __init__(self):
        self.products = []
    def add_product(self,product):
        self.products.append(product)
    def remove_product(self,name):
        for product in self.products:
            if product.name.lower() == name.lower():
                self.products.remove(product)
                return
    def get_total_value(self):
        sum  = 0
        for item in self.products:
            sum += int(item.get_price())
        return sum
    def show_summary(self):
        print(len(self.products))
        print(self.get_total_value())
iventory = Inventory()
iventory.add_product(Product("shoes","2030","sports"))
iventory.add_product(Product("desk","3400","furniture"))
iventory.remove_product("shoes")
iventory.get_total_value()
iventory.show_summary()

class store:
    def __init__(self):
        self.inventory = Inventory()
        self.store_name = ""
    def add_new_product(self):
        name = input("Enter product name")
        price = input("Enter Price")
        category = input("Enter category")
        self.inventory.add_product(Product(name,price,category))
    def show_summary(self):
        sum = 0
        print(len(self.inventory.products))
        for product in self.inventory.products:
            sum += int(product.get_price())
        print(sum)
store = store()
for i in range(3):
    store.add_new_product()
store.show_summary()



  
```
