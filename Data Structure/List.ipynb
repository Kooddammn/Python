## By using this data structure we can apply the crud like inserting the data, deleting updating and reading the data

```python
print(lists[-1]) #print the last value
print(lists[1:3]) #print the value from index 1 to index 2
print(lists[:-1]) #print values from index 0 to second last index value
print(lists[::-1]) #reverse the list [start:stop:step]
```

```python
squares = [x**2 for x in range(5)]
print(squares)
```

```python
even = [x for x in range(10) if x%2==0]
print(even)
```

```python
labels = ["Even" if x%2 ==0 else "Odd" for x in range(10)]
print(labels)
```

```python
pairs = [(x,y) for x in [1,2] for y in [3,4]]
print(pairs)
```

```python
#list comprehension : concise way to create a list with existing iterable
square = [x**2 for x in range(1,10) if x%2==0]
print(square)
word = ['Hello','world','python']
upper = [item.upper() for item in word]
print(upper)
nested = [[i*j for j in range(1,4)] for i in range(1,4)]
print(nested)
```
