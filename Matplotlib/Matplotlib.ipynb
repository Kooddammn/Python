```python
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
axes[0,0].plot([1,2,3], [4,5,6])
axes[0,0].set_title('Top Left')
axes[0,1].scatter([1,2,3], [6,2,8])
axes[0,1].set_title('Top Right')
axes[1,0].bar(['A','B','C'], [3,7,5])
axes[1,0].set_title('Bottom Left')
axes[1,1].hist([1,2,2,3,3,3,4,4,4,4])
axes[1,1].set_title('Bottom Right')
plt.tight_layout() # prevent overlap
plt.show()
```

```python
years = [2018, 2019, 2020, 2021, 2022, 2023]
sales_A = [120, 145, 130, 160, 175, 195]
sales_B = [90, 105, 98, 130, 142, 165]
sales_C = [123,465,864,98,542,876]
plt.figure(figsize=(9, 5))
plt.plot(years, sales_A, marker='o', label='Product A', color='steelblue')
plt.plot(years, sales_B, marker='s', label='Product B', color='tomato')
plt.plot(years, sales_C, marker='^', label='Product C', color='purple')
plt.title('Annual Sales Comparison')
plt.xlabel('Year')
plt.ylabel('Sales (units)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
np.random.seed(10)
n = 50
x = np.random.rand(n)
y = np.random.rand(n)
sizes = np.random.rand(n) * 500 # bubble size
colors = np.random.rand(n) # colour value
plt.figure(figsize=(8, 6))
sc = plt.scatter(x, y, s=sizes, c=colors, cmap='viridis', alpha=0.7)
plt.colorbar(sc, label='Colour Scale') # add colour bar
plt.title('Bubble Chart')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
print(sizes)
print(x)
```

```python
m, b = np.polyfit(x, y, 1) # degree-1 polynomial (line)
print(m)
print(b)
trend = m * x + b
plt.scatter(x, y, alpha=0.6, label='Data')
plt.plot(np.sort(x), m * np.sort(x) + b,
color='red', linewidth=2, label=f'y = {m:.2f}x + {b:.2f}')
plt.legend()
plt.title('Scatter with Trend Line')
plt.show()
```

```python
categories = ['Python', 'JavaScript', 'Java', 'C++', 'Go']
popularity = [30.3, 13.5, 17.8, 10.2, 5.1]
plt.figure(figsize=(8, 5))

bars = plt.bar(categories, popularity, color='steelblue', edgecolor='navy',
width=0.6)
# Add value labels on top of each bar
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.3,
    f'{height}%', ha='center', va='bottom', fontsize=10)
plt.title('Programming Language Popularity')
plt.xlabel('Language')
plt.ylabel('Popularity (%)')
plt.ylim(0, 35)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
subjects = ['Maths', 'Science', 'English', 'History']
student_A = [85, 92, 78, 88]
student_B = [76, 84, 90, 72]
student_C = [91, 88, 82, 95]
x = np.arange(len(subjects))
print(x)
width = 0.25
plt.figure(figsize=(9, 5))
plt.bar(x - width, student_A, width, label='Alice', color='steelblue')
plt.bar(x, student_B, width, label='Bob', color='tomato')
plt.bar(x + width, student_C, width, label='Charlie', color='seagreen')
plt.xticks(x, subjects)
plt.xlabel('Subject')
plt.ylabel('Score')
plt.title('Student Score Comparison')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Sample data
categories = ['Q1', 'Q2', 'Q3', 'Q4']
group_a = [20, 35, 30, 35]
group_b = [25, 32, 34, 20]

fig, ax = plt.subplots()

# Plot first layer
ax.bar(categories, group_a, label='Group A', color='#4f46e5')

# Plot second layer on top of group_a
ax.bar(categories, group_b, bottom=group_a, label='Group B', color='#06b6d4')

ax.set_ylabel('Scores')
ax.set_title('Stacked Bar Plot Example')
ax.legend()

plt.show()
```
