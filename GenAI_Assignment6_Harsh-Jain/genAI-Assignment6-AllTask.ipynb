```python
#Task 1: Safe Division Utility
try:
 num = int(input("Enter numerator"))
 deno = int(input("Enter denominator"))
except ValueError as vexception:
    print(f'the input should be a a number : {vexception}')

try:
    div = num/deno
except ZeroDivisionError as zexception:
    print(zexception)
else:
    print(div)
finally:
    print("Operation Complete")
```

```python
#Task 2: Billl calculator with Error Handling
prices = [120, 350, 'abc', 500, -200, 800]

total = 0

for price in prices:
    try:
        
        if not isinstance(price, (int, float)):
            raise TypeError("Price must be a number")

        
        if price < 0:
            raise ValueError("Negative price not allowed")

        total += price
        print(f"Added {price}, Running Total = {total}")

    except TypeError as e:
        print(f"Type Error: {e} -> {price}")

    except ValueError as e:
        print(f"Value Error: {e} -> {price}")

print(f"\nFinal Total = {total}")
```

```python
#Task 3: Custom Exception : Age Validator
def check_age(age):
    if age < 1 or age > 120:
        raise ValueError("Age must be between 1 and 120")
    return True

try:
    age = int(input("Enter your age: "))

    check_age(age)

    print(f"Valid age entered: {age}")

except ValueError as e:
    print(f"Error: {e}")     
```

```python
#Task 4: File reader with exception handling
try:
    filename = input("Enter the filename: ")

    with open(filename, "r") as file:
        print("\nFirst 3 lines of the file:")

        for i in range(3):
            line = file.readline()

           
            if not line:
                break

            print(line.strip())

except FileNotFoundError:
    print("Error: File not found.")

except PermissionError:
    print("Error: Permission denied. Cannot access the file.")

finally:
    print("File operation attempted.")
```

```python
#Task 5: Mini program - Safe shopping cart
cart = []

while True:
    try:
        user_input = input("Enter item price (or 'q' to quit): ")

        
        if user_input.lower() == 'q':
            break

        price = float(user_input)

        if price < 0:
            raise ValueError("Negative price is not allowed")

        cart.append(price)
        print(f"Added ₹{price:.2f} to cart")

    except ValueError as e:
        print(f"Error: {e}")

total_items = len(cart)
total_bill = sum(cart)

print("\n----- Shopping Cart Summary -----")
print(f"Total Items: {total_items}")
print(f"Total Bill: ₹{total_bill:.2f}")
    
```
