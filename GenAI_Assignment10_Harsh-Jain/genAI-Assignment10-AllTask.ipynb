```python
#Task 1: Pandas Series Basics
import pandas as pd
marks = [78,85,90,66,72]

s = pd.Series(marks)
print(s.values)
print(s.index)
print(s.dtype)
print(s[0])
print(s[len(s) - 2 :])
```

```python
#Task 2: Mathematical Operations on Series
#1. Add 5 grace marks to all student
print(s.add(5, fill_value=0))
#2. Subtract 2 from all marks
print(s.subtract(2, fill_value=0))
#3. Multiply all mark by 1.05
print(s.multiply(1.05, fill_value=0))
#4. Divide all marks by 2
print(s.divide(2, fill_value=0))
```

```python
#Task 3: Python functionality on series
print(s.max())
print(s.min())
print(s.sum())
print(s.mean())

#2. Apply lambda function to check whether each student has passed
checkPassed = lambda _marks : _marks >= 70
print(checkPassed(s))
#3.Count how many student passed
print(len(list(filter(lambda x : x== True,checkPassed(s)))))
```

```python
#Task 4: Create a Data Frame
students ={
    'Names':['Amit','Neha','Rahul','Sneha','Pooja'],
    'Marks':[78,85,90,66,72],
    'Subject':['Math','Math','Science','Science','Math']
}
#1.Convert it into dataframe
df = pd.DataFrame(students)
#2.Print first 3 rows
print(df.head(3))
#3.Print last 2 rows
print(df.tail(2))
#4.Print dataframe shape and column name
print(df.shape)
print(df.columns)
```

```python
#Task 5:Important Dataframe functions
#1.Use and print output of
#info
print(df.info())
#describe
print(df.describe())
#head
print(df.head)
#tail
print(df.tail)
#2. Sort student by marks in descending order
sort =df.sort_values('Marks', ascending=False)
print(sort)
#3. Reset Index after sorting
print(sort.reset_index)
```

```python
#Task 6: Filtering and Conditional Selection
#1. Student who score more than 75 marks
print(df[df['Marks'] > 75])
#2. Student belonging to subject math
print(df[df['Subject'] == 'Math'])
#3. Student who score more than average marks
avg = df['Marks'].mean()
print(df[df['Marks'] > avg])
#4. student who fail(marks<70)
print(df[df['Marks'] < 70])
```

```python
#Task 7: Grouping and Basics Analysis
#1. Find average marks per subject 
print(df.groupby('Subject')['Marks'].mean())
#2. Count number of student per subject
print(df.groupby('Subject')['Names'].count())
#3. Find maximum marks per subject
print(df.groupby('Subject')['Marks'].max())
```

```python
#Task 8: Pandas Plotting(Simple graph)
#1.
import matplotlib.pyplot as plt
s = pd.Series(df.Marks.values, index = df['Names'])
s.plot(kind='bar',color='blue')
plt.show()
#2.
s.plot(kind='line',color='blue')
plt.show()
#3.
s.plot(kind='hist',color='blue')
plt.show()
```

```python
#Task 9: Mini use case - Sales data analysis 
sale ={
    'Days':['Mon','Tue','Wed','Thu','Fri'],
    'Revenue':[1200,1500,900,2000,1800]
}
df_sale = pd.DataFrame(sale)
#1. Total Revenue
print(df_sale['Revenue'].sum())
#2. Average daily revenue
avg = df_sale['Revenue'].mean()
print(avg)
#3. Day with highest revenue
sorted = df_sale.sort_values('Revenue', ascending=False)
print(sorted.Days.head(1))
#4. Days where revenue greater than average
print(df_sale.Days[df_sale.Revenue > avg])
#5 . Plot revenue vs Days
s = pd.Series(df_sale.Revenue.values, index=df_sale['Days'])
s.plot(kind='bar', color='blue')
```
