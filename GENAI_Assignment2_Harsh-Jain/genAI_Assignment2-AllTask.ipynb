```python
#Task 1: Discount Rule
user_input = input("Enter the order amount")
try:
    order_amount = int(user_input)
except ValueError:
    print("User input is not valid amount")
if order_amount >= 2000:
    order_amount = order_amount - (order_amount * (0.15))
elif order_amount >= 1500 and order_amount < 2000:
    order_amount = order_amount - (order_amount * (0.10))
elif order_amount >= 1000 and order_amount < 1500:
    order_amount = order_amount - (order_amount * 0.07)
else:
    order_amount
print(order_amount)
```

```python
#Task 2: Process Multiple orders
orders = [1200,2500,800,1750,3000]
final_amount = 0
discount = 0
summary_table =[()]
total_revenue = 0;
for order_price in orders:
    if order_price >= 2000:
        final_amount = order_price - (order_price * (0.15))
        discount = 15
    elif order_price >= 1500 and order_price < 2000:
        final_amount = order_price - (order_price * (0.10))
        discount = 10
    elif order_price >= 1000 and order_price < 1500:
        final_amount = order_price - (order_price * 0.07)
        discount = 7
    else:
        final_amount = order_price
        discount = 0
    summary_table.append([order_price,discount,final_amount])
print(summary_table[1:])
for order_price,discount,final_amount in summary_table[1:]:
     total_revenue += final_amount
print(total_revenue)
```

```python
# Task 3: User Menu (while loop + break/continue)
while(True):
    user_input = input("Choose an action")
    if user_input not in ["1","2","q"]:
        continue
    if user_input == "1":
       order_amt = input("Enter order amount value")
       order_item = int(order_amt)
       orders.append(order_item)
       print(orders)
       continue
    elif user_input == "2":
        if order_item >= 2000:
            final_amount = order_item - (order_item * (0.15))
            discount = 15
        elif order_item >= 1500 and order_item < 2000:
            final_amount = order_item - (order_item * (0.10))
            discount = 10
        elif order_item >= 1000 and order_item < 1500:
            final_amount = order_item - (order_item * 0.07)
            discount = 7
        else:
            final_amount = order_item
            discount = 0  
        total_revenue += final_amount
        print(total_revenue)  
        continue       
    elif user_input == "q":
        break
    
```

```python
#Task 4: Loop control with conditions(break/continue)
daily = [200, 150, 0, 400, 50, -1, 300]
total_sales = 0
for item in daily:
    if item == -1:
        print(total_sales)
        break
    elif item == 0:
        continue
    else:
        total_sales += item
        print(total_sales)
print(total_sales)
```
