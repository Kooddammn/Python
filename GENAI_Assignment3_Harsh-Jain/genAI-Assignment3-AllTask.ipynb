```python
#Task 1: Basic Function: Price after discount
def apply_discount(price, discount_percent = 5):
    if discount_percent > 60:
        return price
    return price - (price * (discount_percent)/100)
    
print(apply_discount(1000,10))
print(apply_discount(500))
print(apply_discount(1000, 65))
```

```python
#Task 2: Recursive function - Factorial Utility
def factorial(n):
    if n < 0:
        print("Number should be positive")
        return
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)
print(factorial(5))
print(factorial(0))
print(factorial(-3))
```

```python
#Task 3: Lambda function - GST calculator
gst = lambda price : price + (0.18 * price)
final_price = lambda f_price : f_price + (0.18 * f_price) - (0.05 * f_price)
print(gst(100))
print(final_price(200))
```

```python
#Task 4: Using map(): Apply GST to list of price
prices = [100, 250, 400, 1200, 50]
price_with_gst = list(map(lambda x : x + (0.18 * x), prices))
print(prices)
print(price_with_gst)
```

```python
#Task 5 : Using filter() : filter expensive product
prices = [100, 250, 400, 1200, 50, 2000, 850]
filter_greater_price = list(filter(lambda n : n > 500,prices))
filter_lesser_price = list(filter(lambda n : n <= 500, prices))
print(filter_greater_price)
print(filter_lesser_price)
```

```python
#Task 6 : Combined Utility function
def process_price(prices):
    discounted_price = list(map(lambda price : price - (0.10 * price), prices))
    filtered_prices = list(filter(lambda n : n > 300, discounted_price))
    return discounted_price,filtered_prices
print(process_price([100,500,900,50,750]))
```

```python
#Task 7: Mini Problem - Menu using function
prices_list = [12,45,76,43]
def add_prices(prices_list, price):
    prices_list.append(price)
    return prices_list
get_average_price = lambda n: sum(n)/len(prices_list)
get_max_price = lambda n : max(n)
print(add_prices(prices_list,65))
print(get_average_price(prices_list))
print(get_max_price(prices_list))
```
